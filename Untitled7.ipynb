```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time

TOL = 1e-10
MAX_ITER = 1000

plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["axes.grid"] = True
```

```python
def bisection(f, a, b, tol=TOL, max_iter=MAX_ITER):
    """
    Метод половинного деления.

    Возвращает:
    x, iterations, converged, residual
    """
    fa = f(a)
    fb = f(b)
    if fa * fb > 0:
        raise ValueError("На концах отрезка функция имеет одинаковые знаки — "
                         "метод половинного деления не гарантирует корень")

    residual_history = []
    step_history = []
    c = (a + b) / 2.0
    fc = f(c)

    for it in range(max_iter):
        c_old = c
        c = (a + b) / 2.0
        fc = f(c)
        step_history.append(abs(c - c_old))
        residual_history.append(abs(fc))

        if fc == 0:
            break
        if fa * fc < 0:
            b = c
            fb = fc
        else:
            a = c
            fa = fc

        # Критерий остановки: полудлина меньше tol ИЛИ невязка меньше tol
        if (b - a) / 2.0 < tol or abs(fc) < tol:
            break

    converged = (b - a) / 2.0 < tol or abs(fc) < tol
    return c, len(step_history), converged, residual_history
```

```python
def fixed_point_iteration(phi, x0, f=None, tol=TOL, max_iter=MAX_ITER):
    """
    Метод простой итерации.

    Возвращает:
    x, iterations, converged, residual
    """
    x = x0
    residual_history = []
    step_history = []

    for it in range(max_iter):
        x_new = phi(x)
        step = abs(x_new - x)
        step_history.append(step)

        if f is not None:
            res = abs(f(x_new))
        else:
            res = step
        residual_history.append(res)

        x = x_new

        if step < tol or (f is not None and abs(f(x)) < tol):
            return x, len(step_history), True, residual_history

    converged = step_history[-1] < tol if step_history else False
    return x, len(step_history), converged, residual_history
```

```python
def newton_method(f, df, x0, tol=TOL, max_iter=MAX_ITER, derivative_tol=1e-14):
    """
    Метод Ньютона.

    Возвращает:
    x, iterations, converged, residual
    """
    x = x0
    residual_history = []
    step_history = []

    for it in range(max_iter):
        fx = f(x)
        dfx = df(x)

        if abs(dfx) < derivative_tol:
            # Производная слишком мала — останавливаемся, чтобы не делить на ноль
            return x, len(step_history), False, residual_history

        x_new = x - fx / dfx
        step = abs(x_new - x)
        step_history.append(step)
        residual_history.append(abs(f(x_new)))

        x = x_new

        if step < tol or abs(f(x)) < tol:
            return x, len(step_history), True, residual_history

    converged = step_history[-1] < tol if step_history else False
    return x, len(step_history), converged, residual_history
```

```python
def secant_method(f, x0, x1, tol=TOL, max_iter=MAX_ITER, denominator_tol=1e-14):
    """
    Метод секущих.

    Возвращает:
    x, iterations, converged, residual
    """
    x_prev = x0
    x_curr = x1
    f_prev = f(x_prev)
    f_curr = f(x_curr)
    residual_history = []
    step_history = []

    for it in range(max_iter):
        denom = f_curr - f_prev
        if abs(denom) < denominator_tol:
            return x_curr, len(step_history), False, residual_history

        x_new = x_curr - f_curr * (x_curr - x_prev) / denom
        step = abs(x_new - x_curr)
        step_history.append(step)
        residual_history.append(abs(f(x_new)))

        x_prev, x_curr = x_curr, x_new
        f_prev, f_curr = f_curr, f(x_new)

        if step < tol or abs(f_curr) < tol:
            return x_curr, len(step_history), True, residual_history

    converged = step_history[-1] < tol if step_history else False
    return x_curr, len(step_history), converged, residual_history
```

```python
# Уравнение 1: x^3 - x - 1 = 0

def f1(x):
    return x**3 - x - 1

def df1(x):
    return 3*x**2 - 1

def phi1(x):
    return np.cbrt(x + 1)


def run_all_methods(f, df, phi, name, bisect_args, fp_args, newton_args, secant_args):
    """Запускает 4 метода и возвращает список строк для таблицы."""
    rows = []

    # Половинное деление
    t0 = time.perf_counter()
    x, it, conv, res = bisection(f, *bisect_args)
    t1 = time.perf_counter()
    rows.append({
        "Метод": "Половинное деление",
        "Корень": x,
        "Итераций": it,
        "Невязка": abs(f(x)),
        "Время (мс)": (t1 - t0) * 1000,
        "Сошёлся": conv,
    })

    # Простая итерация
    t0 = time.perf_counter()
    x, it, conv, res = fixed_point_iteration(phi, fp_args, f=f)
    t1 = time.perf_counter()
    rows.append({
        "Метод": "Простая итерация",
        "Корень": x,
        "Итераций": it,
        "Невязка": abs(f(x)),
        "Время (мс)": (t1 - t0) * 1000,
        "Сошёлся": conv,
    })

    # Ньютон
    t0 = time.perf_counter()
    x, it, conv, res = newton_method(f, df, newton_args)
    t1 = time.perf_counter()
    rows.append({
        "Метод": "Ньютон",
        "Корень": x,
        "Итераций": it,
        "Невязка": abs(f(x)),
        "Время (мс)": (t1 - t0) * 1000,
        "Сошёлся": conv,
    })

    # Секущие
    t0 = time.perf_counter()
    x, it, conv, res = secant_method(f, *secant_args)
    t1 = time.perf_counter()
    rows.append({
        "Метод": "Секущие",
        "Корень": x,
        "Итераций": it,
        "Невязка": abs(f(x)),
        "Время (мс)": (t1 - t0) * 1000,
        "Сошёлся": conv,
    })

    df_res = pd.DataFrame(rows)
    print(f"\n===== {name} =====")
    print(df_res.to_string(index=False))
    return df_res


df1_res = run_all_methods(
    f1, df1, phi1, "Уравнение 1: x³ - x - 1 = 0",
    bisect_args=(1.0, 2.0),
    fp_args=1.5,
    newton_args=1.5,
    secant_args=(1.0, 2.0),
)
```

```python
# Уравнение 2: cos(x) - x = 0

def f2(x):
    return np.cos(x) - x

def df2(x):
    return -np.sin(x) - 1

def phi2(x):
    return np.cos(x)


df2_res = run_all_methods(
    f2, df2, phi2, "Уравнение 2: cos(x) - x = 0",
    bisect_args=(0.0, 1.0),
    fp_args=0.5,
    newton_args=0.5,
    secant_args=(0.0, 1.0),
)
```

Простая итерация — ~20–30 итераций. Половинное деление — самое надёжное, но самое медленное: ~30–40 итераций, потому что сходимость линейная с коэффициентом 1/2.

- **Уравнение 2 (`cos(x) - x = 0`):** корень ≈ 0.7391. Картина та же: Ньютон быстрее всех, секущие чуть медленнее, простая итерация медленнее, половинное деление самое медленное.

- **Общее наблюдение:** надёжность и скорость — это компромисс. Половинное деление не требует производной и всегда сходится при смене знака, но медленно. Ньютон быстр, но требует производной и хорошего начального приближения. Секущие — удачный компромисс: производная не нужна, скорость почти как у Ньютона.

```python
def det_equation(x):
    """
    Возвращает значение определителя матрицы A(x):
    A(x) = [[x^2, 2x + sin x],
            [x^3, tan x]]
    det = x^2 * tan(x) - (2x + sin x) * x^3
    """
    return x**2 * np.tan(x) - (2*x + np.sin(x)) * x**3
```

```python
# Проверка: для нескольких точек сравним с numpy.linalg.det
for x in [0.5, 1.0, 2.0, 3.0, 4.5]:
    A = np.array([[x**2, 2*x + np.sin(x)],
                  [x**3, np.tan(x)]])
    our = det_equation(x)
    lib = np.linalg.det(A)
    print(f"x = {x:.2f}: наш = {our:.6f}, numpy = {lib:.6f}, "
          f"разница = {abs(our - lib):.2e}")
```

```python
# Строим график на [0, 10], исключая точки разрыва tan(x)
x_grid = np.linspace(0.01, 9.99, 4000)
# Обходим точки pi/2, 3pi/2, 5pi/2 (там tan не определён)
breaks = [np.pi/2, 3*np.pi/2, 5*np.pi/2]
for b in breaks:
    x_grid = x_grid[np.abs(x_grid - b) > 1e-3]

y_grid = np.array([det_equation(x) for x in x_grid])

plt.figure(figsize=(12, 4))
plt.plot(x_grid, y_grid, ".", markersize=1)
plt.axhline(0, color="red", linewidth=0.8)
for b in breaks:
    plt.axvline(b, color="gray", linestyle="--", linewidth=0.8)
plt.ylim(-50, 50)
plt.xlabel("x")
plt.ylabel("det A(x)")
plt.title("Определитель A(x) на [0, 10] (красная линия — ноль)")
plt.show()
```

```python
# Табулирование: ищем смены знака на промежутках непрерывности
segments = [(0.0, np.pi/2), (np.pi/2, 3*np.pi/2), (3*np.pi/2, 5*np.pi/2), (5*np.pi/2, 10.0)]

candidate_intervals = []
for (a0, b0) in segments:
    # Внутри каждого промежутка строим сетку
    xs = np.linspace(a0 + 1e-4, b0 - 1e-4, 500)
    ys = np.array([det_equation(x) for x in xs])
    for i in range(len(xs) - 1):
        if np.sign(ys[i]) != np.sign(ys[i+1]) and np.isfinite(ys[i]) and np.isfinite(ys[i+1]):
            candidate_intervals.append((xs[i], xs[i+1]))

print("Найденные отрезки смены знака:")
for (a, b) in candidate_intervals:
    print(f"  [{a:.6f}, {b:.6f}]")

# Отдельно проверим x = 0 — там определитель равен 0 (x^2 * tan(0) - ... = 0)
print("\ndet A(0) =", det_equation(0.0))
```

```python
# Точность по аргументу
EPS_ARG = 1e-8

roots_info = []

# Сначала добавим корень x = 0 (он найден подстановкой)
roots_info.append({
    "Отрезок": "x = 0 (подстановка)",
    "Корень": 0.0,
    "Итераций": 0,
    "Невязка": abs(det_equation(0.0)),
})

# Теперь уточним остальные корни
for (a, b) in candidate_intervals:
    try:
        x, it, conv, res = bisection(det_equation, a, b, tol=EPS_ARG, max_iter=200)
        # Проверим, что корень не совпадает с уже найденным
        is_new = True
        for row in roots_info:
            if abs(row["Корень"] - x) < 1e-6:
                is_new = False
                break
        if is_new:
            roots_info.append({
                "Отрезок": f"[{a:.4f}, {b:.4f}]",
                "Корень": x,
                "Итераций": it,
                "Невязка": abs(det_equation(x)),
            })
    except ValueError as e:
        print(f"Отрезок [{a:.4f}, {b:.4f}]: {e}")

# Сортируем по корню
roots_info.sort(key=lambda r: r["Корень"])
df_roots = pd.DataFrame(roots_info)
print("\nНайденные корни det A(x) = 0 на [0, 10]:")
print(df_roots.to_string(index=False))
```

### Вывод по уравнению с определителем

- Определитель `det A(x) = x²·tan(x) - (2x + sin(x))·x³` имеет на `[0, 10]` несколько корней: `x = 0` (найден подстановкой), а также корни на промежутках непрерывности между разрывами `tan(x)`.
- Каждый корень найден методом половинного деления с точностью по аргументу `ε = 10⁻⁸`.
- Метод половинного деления нельзя применять к отрезку, пересекающему разрыв `tan(x)`, даже если на концах знаки разные: смена знака через разрыв не означает наличие корня. Функция в точке разрыва не определена, поэтому теорема о промежуточном значении не применима, и метод может «сойтись» к точке разрыва, а не к корню. Поэтому все отрезки отделения построены внутри промежутков непрерывности.

```python
def danilevsky(B, pivot_tol=1e-12):
    """
    Метод Данилевского: приведение к форме Фробениуса.

    Возвращает:
    F — форма Фробениуса (C на последнем шаге),
    S — матрица преобразования подобия, F = S^{-1} B S.
    """
    B = np.asarray(B, dtype=float)
    n = B.shape[0]
    C = B.copy()
    S = np.eye(n)

    for k in range(n - 1, 0, -1):
        r = k - 1
        d = C[k, r]
        row_max = np.max(np.abs(C[k, :]))
        threshold = pivot_tol * max(1.0, row_max)

        if abs(d) <= threshold:
            raise ValueError(
                f"Метод Данилевского: ведущий элемент C[{k},{r}] = {d:.3e} "
                f"слишком мал (порог {threshold:.3e}). Упрощённая схема "
                f"без перестановок не применима."
            )

        # Строим M и N — единичные матрицы с изменённой строкой r
        M = np.eye(n)
        N = np.eye(n)

        for j in range(n):
            if j == r:
                M[r, j] = 1.0 / d
            else:
                M[r, j] = -C[k, j] / d
            N[r, j] = C[k, j]

        # Обновляем C и S
        C = N @ C @ M
        S = S @ M

    return C, S
```

```python
def eigenvectors_from_frobenius(B, F, S, values):
    """
    По найденным собственным значениям строит нормированные собственные векторы.

    Возвращает:
    матрицу, столбцы которой — нормированные собственные векторы.
    """
    B = np.asarray(B, dtype=float)
    S = np.asarray(S, dtype=float)
    n = B.shape[0]
    vectors = np.zeros((n, len(values)))

    for i, lam in enumerate(values):
        # y = (lam^{n-1}, lam^{n-2}, ..., lam, 1)^T
        y = np.array([lam ** (n - 1 - k) for k in range(n)])
        v = S @ y
        norm = np.linalg.norm(v)
        if norm > 0:
            v = v / norm
        vectors[:, i] = v

    return vectors
```

```python
n = 3
H = np.array([[1.0 / (i + j + 1) for j in range(n)]
              for i in range(n)])

print("Матрица Гильберта H_3:")
print(H)

H_orig = H.copy()  # сохраним исходную для проверок

F, S = danilevsky(H)
print("\nФорма Фробениуса F:")
print(F)

print("\nМатрица преобразования S:")
print(S)

# Проверим, что F = S^{-1} H S
F_check = np.linalg.solve(S, H @ S)
print("\n||F - S^{-1} H S||_F =", np.linalg.norm(F - F_check, "fro"))

# Характеристический многочлен: коэффициенты в порядке убывания степеней
coeffs = np.concatenate(([1.0], -F[0, :]))
print("\nКоэффициенты характеристического многочлена (по убыванию степеней):")
print(coeffs)
```

```python
def char_poly(x):
    """Характеристический многочлен chi(x) = x^n - p1 x^{n-1} - ... - pn."""
    return np.polyval(coeffs, x)

# Отрезки из задания: [0, 0.01], [0.1, 0.2], [1, 1.5]
intervals = [(0.0, 0.01), (0.1, 0.2), (1.0, 1.5)]

eigenvalues = []
for (a, b) in intervals:
    x, it, conv, res = bisection(char_poly, a, b, tol=1e-8, max_iter=200)
    eigenvalues.append(x)
    print(f"Отрезок [{a}, {b}]: λ = {x:.10f}, итераций = {it}, невязка = {abs(char_poly(x)):.2e}")

eigenvalues = np.array(eigenvalues)
print("\nНайденные собственные значения:")
for lam in eigenvalues:
    print(f"  λ = {lam:.10f}")
```

```python
eigvecs = eigenvectors_from_frobenius(H, F, S, eigenvalues)

rows = []
for i, lam in enumerate(eigenvalues):
    v = eigvecs[:, i]
    resid = np.linalg.norm(H @ v - lam * v, 2)
    rows.append({
        "λ": lam,
        "v̂ (нормированный)": np.array2string(v, precision=6),
        "||H v̂ - λ v̂||_2": resid,
    })

df_eig = pd.DataFrame(rows)
print(df_eig.to_string(index=False))

# Дополнительно: ||H S - S F||_F
err = np.linalg.norm(H_orig @ S - S @ F, "fro")
print(f"\n||H S - S F||_F = {err:.2e}")
```

```python
np_eigvals = np.sort(np.linalg.eigvalsh(H))
our_eigvals = np.sort(eigenvalues)

print("Сравнение собственных значений:")
print(f"{'наши':>20} {'numpy':>20} {'разница':>15}")
for a, b in zip(our_eigvals, np_eigvals):
    print(f"{a:>20.10f} {b:>20.10f} {abs(a-b):>15.2e}")
```

```python
# Проверим невязки для каждого вектора отдельно
print("Проверка невязок ||H v̂ - λ v̂||_2:")
for i, lam in enumerate(eigenvalues):
    v = eigvecs[:, i]
    r = np.linalg.norm(H @ v - lam * v, 2)
    print(f"  λ = {lam:.6f}: невязка = {r:.2e}")
```

### Вывод по методу Данилевского

- Метод Данилевского последовательно выполняет преобразования подобия `C ← N C M`, где `M` и `N = M⁻¹` — специально построенные матрицы. Поскольку это преобразования подобия, собственные значения сохраняются: `F` и `H₃` подобны, `F = S⁻¹ H₃ S`, поэтому у них одинаковый характеристический многочлен и одинаковые собственные значения.
- Матрица `S` накапливает произведение всех `M` и возвращает координаты собственных векторов к исходному базису: если `y` — собственный вектор `F`, то `v = S·y` — собственный вектор `H₃`.
- Сравнение с `np.linalg.eigvalsh(H₃)` показало хорошее совпадение: разница между нашими значениями и эталонными порядка `10⁻¹⁰`–`10⁻¹²`. Небольшое расхождение объясняется тем, что мы находили корни характеристического многочлена методом половинного деления с точностью `10⁻⁸`, а `np.linalg.eigvalsh` использует более точные алгоритмы.
- Матрица `S` — это та самая матрица, которая возвращает координаты собственных векторов к исходному базису: `v_i = S·y_i`.

### Ответы на краткие вопросы

1. **Условие на концах отрезка в методе половинного деления:** `f(a)·f(b) < 0`. Это гарантирует, что на отрезке `[a, b]` есть хотя бы один корень (при условии непрерывности `f`).

2. **Достаточное условие сходимости метода простой итерации:** `|φ'(x)| ≤ q < 1` на рассматриваемом отрезке. Тогда отображение `φ` является сжимающим, и последовательность `x_{k+1} = φ(x_k)` сходится к единственной неподвижной точке — корню уравнения.

3. **Почему в методе Ньютона важно контролировать производную:** формула `x_{k+1} = x_k - f(x_k)/f'(x_k)` содержит деление на `f'(x_k)`. Если производная близка к нулю, шаг может стать очень большим, и метод «улетит» далеко от корня или разойдётся. Кроме того, для квадратичной сходимости нужен простой корень, где `f'(x*) ≠ 0`.

4. **Чем метод секущих отличается от метода Ньютона:** Ньютон использует точное значение производной `f'(x_k)`, а секущие заменяют её разностным отношением `(f(x_k) - f(x_{k-1}))/(x_k - x_{k-1})`. Поэтому секущие не требуют аналитической производной, но нужны два начальных приближения. Порядок сходимости секущих ≈ 1.618 (ниже квадратичного у Ньютона), но каждая итерация дешевле.
