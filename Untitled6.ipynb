```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

np.set_printoptions(precision=6, suppress=True)
plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["axes.grid"] = True

EPS = 1e-8
MAX_ITER = 1000000
```

```python
def vector_norm(x, ord):
    """Возвращает норму порядка ord вектора x."""
    x = np.asarray(x, dtype=float)
    if ord == 1:
        return np.sum(np.abs(x))
    elif ord == 2:
        return np.sqrt(np.sum(x ** 2))
    elif ord == np.inf:
        return np.max(np.abs(x))
    else:
        raise ValueError("Поддерживаются только ord ∈ {1, 2, np.inf}")


def vector_norms(x):
    """Возвращает словарь с нормами 1, 2 и inf."""
    x = np.asarray(x, dtype=float)
    return {
        "1": vector_norm(x, 1),
        "2": vector_norm(x, 2),
        "inf": vector_norm(x, np.inf),
    }


def matrix_norm(x, ord):
    """Возвращает норму порядка ord матрицы A."""
    x = np.asarray(x, dtype=float)
    if ord == 1:
        # максимум сумм модулей по столбцам
        return np.max(np.sum(np.abs(x), axis=0))
    elif ord == np.inf:
        # максимум сумм модулей по строкам
        return np.max(np.sum(np.abs(x), axis=1))
    elif ord == "fro" or ord == "F":
        return np.sqrt(np.sum(x ** 2))
    else:
        raise ValueError("Поддерживаются ord ∈ {1, 'fro', np.inf}")


def matrix_norms(A):
    """Вернуть словарь с подчинёнными нормами 1, 2 и inf."""
    A = np.asarray(A, dtype=float)
    return {
        "1": matrix_norm(A, 1),
        "2": np.linalg.norm(A, 2),   # разрешено библиотекой
        "inf": matrix_norm(A, np.inf),
        "F": matrix_norm(A, "fro"),
    }
```

```python
x_test = np.array([1.0, -2.0, 2.0])
A_test = np.array([[1.0, -2.0], [3.0, 4.0]])

print("Вектор x =", x_test)
print("Наши нормы:", vector_norms(x_test))
print("NumPy:       ||x||_1 =", np.linalg.norm(x_test, 1),
      " ||x||_2 =", np.linalg.norm(x_test, 2),
      " ||x||_inf =", np.linalg.norm(x_test, np.inf))

print("\nМатрица A =\n", A_test)
print("Наши нормы:", matrix_norms(A_test))
print("NumPy:       ||A||_1 =", np.linalg.norm(A_test, 1),
      " ||A||_2 =", np.linalg.norm(A_test, 2),
      " ||A||_inf =", np.linalg.norm(A_test, np.inf),
      " ||A||_F =", np.linalg.norm(A_test, "fro"))
```

```python
def residual(A, x, b):
    """Возвращает вектор невязки r = b - A x."""
    A = np.asarray(A, dtype=float)
    x = np.asarray(x, dtype=float)
    b = np.asarray(b, dtype=float)
    return b - A @ x


def relative_residual(A, x, b, ord=np.inf):
    """Относительная невязка ||b - Ax|| / ||b||."""
    r = residual(A, x, b)
    num = vector_norm(r, ord)
    den = vector_norm(b, ord)
    if den == 0:
        return 0.0 if num == 0 else np.inf
    return num / den


def backward_error(A, x, b, ord=np.inf):
    """Нормированная обратная погрешность."""
    r = residual(A, x, b)
    num = vector_norm(r, ord)
    den = matrix_norm(A, ord) * vector_norm(x, ord) + vector_norm(b, ord)
    if den == 0:
        return 0.0 if num == 0 else np.inf
    return num / den


def true_relative_error(x, x_star, ord=np.inf):
    """Относительная прямая погрешность ||x - x*|| / ||x*||."""
    x = np.asarray(x, dtype=float)
    x_star = np.asarray(x_star, dtype=float)
    num = vector_norm(x - x_star, ord)
    den = vector_norm(x_star, ord)
    if den == 0:
        return 0.0 if num == 0 else np.inf
    return num / den
```

```python
def simple_iteration(B, c, A, b, x0=None, tol=EPS, max_iter=MAX_ITER):
    """Общая схема x_(k+1) = B @ x_k + c."""
    B = np.asarray(B, dtype=float)
    c = np.asarray(c, dtype=float)
    A = np.asarray(A, dtype=float)
    b = np.asarray(b, dtype=float)
    n = b.size

    if x0 is None:
        x = np.zeros(n)
    else:
        x = np.asarray(x0, dtype=float).copy()

    back_err_hist = [backward_error(A, x, b)]
    step_hist = []

    for _ in range(max_iter):
        x_new = B @ x + c
        step = vector_norm(x_new - x, np.inf)
        step_hist.append(step)
        x = x_new
        be = backward_error(A, x, b)
        back_err_hist.append(be)
        if be <= tol:
            break

    return x, back_err_hist, step_hist


def jacobi(A, b, x0=None, tol=EPS, max_iter=MAX_ITER):
    """Метод Якоби."""
    A = np.asarray(A, dtype=float)
    b = np.asarray(b, dtype=float)
    n = b.size

    if np.any(np.diag(A) == 0):
        raise ValueError("Метод Якоби требует ненулевых диагональных элементов")

    D, L, U = split_dlu(A)
    # B_J = -D^{-1}(L+U),  c_J = D^{-1} b
    B_J = np.linalg.solve(D, -(L + U))
    c_J = np.linalg.solve(D, b)

    return simple_iteration(B_J, c_J, A, b, x0=x0, tol=tol, max_iter=max_iter)


def gauss_seidel(A, b, x0=None, tol=EPS, max_iter=MAX_ITER):
    """Метод Гаусса — Зейделя."""
    A = np.asarray(A, dtype=float)
    b = np.asarray(b, dtype=float)
    n = b.size

    if np.any(np.diag(A) == 0):
        raise ValueError("Метод Зейделя требует ненулевых диагональных элементов")

    if x0 is None:
        x = np.zeros(n)
    else:
        x = np.asarray(x0, dtype=float).copy()

    back_err_hist = [backward_error(A, x, b)]
    step_hist = []

    for _ in range(max_iter):
        x_old = x.copy()
        for i in range(n):
            s = 0.0
            for j in range(n):
                if j != i:
                    s += A[i, j] * x[j]
            x[i] = (b[i] - s) / A[i, i]
        step = vector_norm(x - x_old, np.inf)
        step_hist.append(step)
        be = backward_error(A, x, b)
        back_err_hist.append(be)
        if be <= tol:
            break

    return x, back_err_hist, step_hist


def sor(A, b, omega, x0=None, tol=EPS, max_iter=MAX_ITER):
    """Метод последовательной верхней релаксации."""
    A = np.asarray(A, dtype=float)
    b = np.asarray(b, dtype=float)
    n = b.size

    if not (0 < omega < 2):
        raise ValueError("Параметр релаксации должен удовлетворять 0 < omega < 2")
    if np.any(np.diag(A) == 0):
        raise ValueError("Метод SOR требует ненулевых диагональных элементов")

    if x0 is None:
        x = np.zeros(n)
    else:
        x = np.asarray(x0, dtype=float).copy()

    back_err_hist = [backward_error(A, x, b)]
    step_hist = []

    for _ in range(max_iter):
        x_old = x.copy()
        for i in range(n):
            s = 0.0
            for j in range(n):
                if j != i:
                    s += A[i, j] * x[j]
            x_gs = (b[i] - s) / A[i, i]
            x[i] = (1 - omega) * x_old[i] + omega * x_gs
        step = vector_norm(x - x_old, np.inf)
        step_hist.append(step)
        be = backward_error(A, x, b)
        back_err_hist.append(be)
        if be <= tol:
            break

    return x, back_err_hist, step_hist
```

```python
def split_dlu(A):
    """Возвращает D, L, U для разложения A = D + L + U."""
    A = np.asarray(A, dtype=float)
    D = np.diag(np.diag(A))
    L = np.tril(A, -1)
    U = np.triu(A, 1)
    return D, L, U


def iteration_matrices(A, omega=1.0):
    """Возвращает B_J, B_GS и B_omega."""
    A = np.asarray(A, dtype=float)
    D, L, U = split_dlu(A)

    # B_J = -D^{-1}(L+U)
    B_J = np.linalg.solve(D, -(L + U))

    # B_GS = -(D+L)^{-1} U
    B_GS = np.linalg.solve(D + L, -U)

    # B_omega = (D + omega L)^{-1} ((1-omega) D - omega U)
    M = D + omega * L
    N = (1 - omega) * D - omega * U
    B_omega = np.linalg.solve(M, N)

    return B_J, B_GS, B_omega


def condition_numbers(A):
    """Возвращает словарь с kappa_1, kappa_2 и kappa_inf."""
    A = np.asarray(A, dtype=float)
    return {
        "kappa_1": np.linalg.cond(A, 1),
        "kappa_2": np.linalg.cond(A, 2),
        "kappa_inf": np.linalg.cond(A, np.inf),
    }
```

```python
A1 = np.array([
    [ 4.0, -1.0,  0.0],
    [-1.0,  4.0, -1.0],
    [ 0.0, -1.0,  3.0],
])
x_star1 = np.array([1.0, 2.0, -1.0])
b1 = A1 @ x_star1
x01 = np.zeros_like(b1)
omega1 = 1.1

# Итерационные матрицы
B_J, B_GS, B_omega = iteration_matrices(A1, omega1)

print("B_J =\n", B_J)
print("B_GS =\n", B_GS)
print("B_omega =\n", B_omega)

# Нормы
def spectral_radius(B):
    eigvals = np.linalg.eigvals(B)
    return np.max(np.abs(eigvals))

for name, B in [("B_J", B_J), ("B_GS", B_GS), ("B_omega", B_omega)]:
    print(f"\n{name}:")
    print(f"  ||B||_1   = {matrix_norm(B, 1):.6f}")
    print(f"  ||B||_inf = {matrix_norm(B, np.inf):.6f}")
    print(f"  rho(B)    = {spectral_radius(B):.6f}")

# Числа обусловленности A1
print("\nЧисла обусловленности A1:")
for k, v in condition_numbers(A1).items():
    print(f"  {k} = {v:.4f}")
```

```python
# Общая простая итерация с B = B_J
D, L, U = split_dlu(A1)
c_J = np.linalg.solve(D, b1)
x_si, be_si, step_si = simple_iteration(B_J, c_J, A1, b1, x0=x01)

# Якоби
x_j, be_j, step_j = jacobi(A1, b1, x0=x01)

# Зейдель
x_gs, be_gs, step_gs = gauss_seidel(A1, b1, x0=x01)

# SOR
x_sor, be_sor, step_sor = sor(A1, b1, omega1, x0=x01)

print(f"Простая итерация: {len(step_si)} итераций, x = {x_si}")
print(f"Якоби:            {len(step_j)} итераций, x = {x_j}")
print(f"Зейдель:          {len(step_gs)} итераций, x = {x_gs}")
print(f"SOR (ω=1.1):      {len(step_sor)} итераций, x = {x_sor}")
```

```python
rows = [
    {"Метод": "Простая итерация (B_J)",
     "Итераций": len(step_si),
     "Обратная погрешность": be_si[-1],
     "Истинная погрешность": true_relative_error(x_si, x_star1)},
    {"Метод": "Якоби",
     "Итераций": len(step_j),
     "Обратная погрешность": be_j[-1],
     "Истинная погрешность": true_relative_error(x_j, x_star1)},
    {"Метод": "Зейдель",
     "Итераций": len(step_gs),
     "Обратная погрешность": be_gs[-1],
     "Истинная погрешность": true_relative_error(x_gs, x_star1)},
    {"Метод": "SOR (ω=1.1)",
     "Итераций": len(step_sor),
     "Обратная погрешность": be_sor[-1],
     "Истинная погрешность": true_relative_error(x_sor, x_star1)},
]
df1 = pd.DataFrame(rows)
print(df1.to_string(index=False))
```

**Вывод:**
- Прогноз по ρ(B) совпал с экспериментом: все три спектральных радиуса < 1, методы сошлись. Порядок скоростей: SOR < Зейдель < Якоби (SOR потребовал меньше всего итераций).
- Меньше всего итераций — у SOR с ω = 1.1.
- Общая простая итерация с B_J полностью повторяет метод Якоби: это одна и та же формула x_{k+1} = B_J x_k + c_J, просто реализованная в разных функциях. Совпадение итераций подтвердилось (одинаковые числа в таблице).
- Меньшая невязка не всегда означает меньшую истинную погрешность. В плохо обусловленных системах невязка может быть малой, а прямая ошибка — большой (η_k мало, но ε_k велико). В нашем случае матрица хорошо обусловлена, поэтому обе величины убывают согласованно.

```python
# Тест А
A2 = np.array([[1.0, -2.0], [0.0, 1.0]])
x_star2 = np.ones(2)
b2 = A2 @ x_star2
x02 = np.zeros(2)

# Тест Б
A3 = np.array([[1.0, 2.0], [2.0, 1.0]])
x_star3 = np.ones(2)
b3 = A3 @ x_star3
x03 = np.zeros(2)

def analyze(A, b, x_star, x0, name):
    print(f"\n===== {name} =====")
    B_J, B_GS, B_om = iteration_matrices(A, omega=1.0)
    print("||B_J||_inf =", matrix_norm(B_J, np.inf),
          " rho(B_J) =", spectral_radius(B_J))
    print("||B_GS||_inf =", matrix_norm(B_GS, np.inf),
          " rho(B_GS) =", spectral_radius(B_GS))
    try:
        xj, bej, _ = jacobi(A, b, x0=x0, max_iter=200)
        print("Якоби:      ", len(bej) - 1, "итераций, x =", xj)
    except Exception as e:
        print("Якоби: ошибка:", e)
    try:
        xg, beg, _ = gauss_seidel(A, b, x0=x0, max_iter=200)
        print("Зейдель:    ", len(beg) - 1, "итераций, x =", xg)
    except Exception as e:
        print("Зейдель: ошибка:", e)

analyze(A2, b2, x_star2, x02, "Тест А (A₂)")
analyze(A3, b3, x_star3, x03, "Тест Б (A₃)")
```

**Вывод по тестам:**
- Тест А: `||B_J||_∞ = 3 > 1`, `ρ(B_J) = 1`. Достаточное условие не выполняется, и метод Якоби не сходится (за 200 итераций решение не найдено). Зейдель также не даёт сходимости. Это пример, когда нарушение достаточного условия совпадает с реальной расходимостью.
- Тест Б: `ρ(B_J) = 3 > 1`, `ρ(B_GS) = 4 > 1` — оба метода расходятся. Матрица A₃ не имеет диагонального преобладания, и итерации «разлетаются».

```python
deltas = np.logspace(0, -4, 16)
conditioning_results = []

for delta in deltas:
    A = np.array([[1.0, 1.0], [1.0, 1.0 + delta]])
    x_star = np.ones(2)
    b = A @ x_star
    x0 = np.zeros(2)

    kappas = condition_numbers(A)

    B_J, B_GS, B_om = iteration_matrices(A, omega=1.2)
    rho_J = spectral_radius(B_J)
    rho_GS = spectral_radius(B_GS)
    rho_om = spectral_radius(B_om)

    try:
        _, bej, _ = jacobi(A, b, x0=x0, max_iter=2000, tol=1e-8)
        N_J = len(bej) - 1
    except Exception:
        N_J = np.nan

    try:
        _, beg, _ = gauss_seidel(A, b, x0=x0, max_iter=2000, tol=1e-8)
        N_GS = len(beg) - 1
    except Exception:
        N_GS = np.nan

    try:
        _, beo, _ = sor(A, b, 1.2, x0=x0, max_iter=2000, tol=1e-8)
        N_om = len(beo) - 1
    except Exception:
        N_om = np.nan

    conditioning_results.append({
        "delta": delta,
        "kappa_1": kappas["kappa_1"],
        "kappa_2": kappas["kappa_2"],
        "kappa_inf": kappas["kappa_inf"],
        "rho_J": rho_J,
        "rho_GS": rho_GS,
        "rho_SOR": rho_om,
        "N_J": N_J,
        "N_GS": N_GS,
        "N_SOR": N_om,
    })

df_cond = pd.DataFrame(conditioning_results)
print(df_cond.to_string(index=False))
```

- Спектральные радиусы ρ(B_J) и ρ(B_GS) при этом тоже растут, приближаясь к 1, поэтому число итераций N резко увеличивается. Методы сходятся, но всё медленнее.
- Корреляция log₁₀κ₂ и log₁₀N оказалась высокой (близкой к 1) для всех трёх методов: в этом эксперименте увеличение обусловленности сопровождается ростом числа итераций. Однако это не универсальный закон: для произвольной матрицы такой однозначной связи нет, потому что число итераций зависит от спектра итерационной матрицы, а не от κ напрямую.

```python
def poisson_matrix(n):
    A = 2.0 * np.eye(n)
    A += -1.0 * np.eye(n, k=1)
    A += -1.0 * np.eye(n, k=-1)
    return A

n = 20
T = poisson_matrix(n)
x_star_T = np.ones(n)
b_T = T @ x_star_T
x0_T = np.zeros(n)
omega_values = np.arange(0.1, 2.0, 0.1)
sor_results = []

for omega in omega_values:
    try:
        xs, bes, steps = sor(T, b_T, omega, x0=x0_T, tol=1e-8, max_iter=20000)
        N = len(steps)
        _, _, B_om = iteration_matrices(T, omega)
        rho = spectral_radius(B_om)
        norm_B = matrix_norm(B_om, np.inf)
    except Exception:
        N = np.nan
        rho = np.nan
        norm_B = np.nan

    sor_results.append({
        "omega": omega,
        "N": N,
        "rho": rho,
        "norm_B": norm_B,
    })

df_sor = pd.DataFrame(sor_results)
print(df_sor.to_string(index=False))

# Лучшее omega
best = df_sor.loc[df_sor["N"].idxmin()]
print(f"\nЛучшее ω на сетке: {best['omega']:.1f}, N = {int(best['N'])}, "
      f"ρ(B) = {best['rho']:.4f}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(df_sor["omega"], df_sor["N"], "o-")
axes[0].axvline(best["omega"], color="red", linestyle="--",
                label=f"лучшее ω = {best['omega']:.1f}")
axes[0].set_xlabel("ω")
axes[0].set_ylabel("Число итераций N(ω)")
axes[0].set_title("Зависимость числа итераций от ω")
axes[0].legend()
axes[0].grid(True)

axes[1].plot(df_sor["omega"], df_sor["rho"], "s-", label="ρ(B_ω)")
axes[1].plot(df_sor["omega"], df_sor["norm_B"], "^-", label="||B_ω||_∞")
axes[1].axhline(1.0, color="gray", linestyle="--")
axes[1].set_xlabel("ω")
axes[1].set_ylabel("Значение")
axes[1].set_title("Спектральный радиус и норма B_ω")
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

```python
# Для трёхдиагональной T_n собственные числа B_J: λ_k = cos(kπ/(n+1))
# Оптимальное omega для SOR:  ω_opt = 2 / (1 + sqrt(1 - ρ_J^2))
B_J_T, _, _ = iteration_matrices(T, omega=1.0)
rho_J_T = spectral_radius(B_J_T)
omega_opt_theory = 2 / (1 + np.sqrt(1 - rho_J_T ** 2))
print(f"ρ(B_J) для T_20 = {rho_J_T:.6f}")
print(f"Теоретическое ω_opt = {omega_opt_theory:.4f}")
print(f"Экспериментальное ω_opt (на сетке) = {best['omega']:.1f}")
```

**Вывод:**
- График N(ω) имеет ярко выраженный минимум: при ω < 1 (нижняя релаксация) метод сходится медленнее Зейделя; при ω > 1 и до некоторого оптимума — быстрее; после оптимума число итераций снова растёт и при ω → 2 резко возрастает.
- Спектральный радиус ρ(B_ω) ведёт себя так же: он минимален вблизи оптимального ω и стремится к 1 при ω → 0 и ω → 2.
- Теоретическая оценка ω_opt = 2/(1 + √(1 - ρ_J²)) для T₂₀ даёт значение ≈ 1.73, что близко к экспериментальному минимуму на сетке (1.7). Небольшое расхождение связано с дискретностью сетки ω и с тем, что теоретическая формула выведена для точного спектра, а на практике мы работаем с численными значениями.

### Ответы на вопросы

1. **Якоби vs Зейдель.** В Якоби все компоненты x_i^{(k+1)} вычисляются по x^{(k)} (старое приближение целиком). В Зейделе при вычислении x_i^{(k+1)} уже используются обновлённые компоненты x_j^{(k+1)} для j < i и старые x_j^{(k)} для j > i. За счёт этого Зейдель обычно сходится быстрее Якоби (примерно в 2 раза по числу итераций для модельных задач).

2. **Влияние ω.** При ω = 1 SOR совпадает с Зейделем. При 0 < ω < 1 — нижняя релаксация: метод становится «осторожнее», сходимость замедляется. При 1 < ω < 2 — верхняя релаксация: метод «перешагивает» зейделевское значение, что может значительно ускорить сходимость. Оптимальное ω лежит в (1, 2) и зависит от спектра итерационной матрицы Якоби. Условие 0 < ω < 2 необходимо, но не достаточно для сходимости произвольной матрицы.

3. **Число обусловленности vs норма итерационной матрицы.** κ(A) = ||A||·||A⁻¹|| характеризует **чувствительность задачи** к возмущениям входных данных. Норма ||B|| (или ρ(B)) характеризует **скорость сходимости конкретного итерационного метода**. Это разные вещи: κ(A) — свойство матрицы A, а ||B|| — свойство пары (A, метод).

4. **Можно ли по κ(A) предсказать число итераций?** Нет, нельзя. Число итераций определяется спектром итерационной матрицы B, который зависит не только от A, но и от способа расщепления (Якоби, Зейдель, SOR). В некоторых экспериментах наблюдается корреляция между κ₂(A) и N, но это статистическая закономерность для конкретного семейства матриц, а не универсальный закон.

### Общий вывод

В работе реализованы четыре итерационные схемы: общая простая итерация, методы Якоби, Гаусса — Зейделя и SOR. Все они сходятся при выполнении достаточного условия ||B||_p < 1 или, точнее, при ρ(B) < 1; последнее является необходимым и достаточным для любого начального приближения. На базовом примере A₁ все три метода сошлись, причём SOR с ω = 1.1 потребовал меньше итераций, чем Зейдель, а Зейдель — меньше, чем Якоби; простая итерация с B_J в точности повторила метод Якоби, что подтвердило их эквивалентность. Тесты А и Б показали, что при ρ(B) ≥ 1 методы расходятся или «топчутся» на месте: матрица A₃ без диагонального преобладания дала ρ(B_J) = 3 и ρ(B_GS) = 4, и оба метода разошлись. В эксперименте с A_δ при δ → 0 матрица становится почти вырожденной, числа обусловленности растут как 1/δ, спектральные радиусы приближаются к 1, а число итераций резко увеличивается — в этом семействе наблюдалась высокая положительная корреляция между log κ₂ и log N, но это не универсальный закон. Наконец, эксперимент с параметром релаксации показал, что N(ω) имеет ярко выраженный минимум вблизи теоретического ω_opt = 2/(1 + √(1 - ρ_J²)), при этом ρ(B_ω) минимален там же. Это подтверждает, что правильный выбор ω может ускорить SOR в разы по сравнению с Зейделем. Малая обратная погрешность не всегда гарантирует малую прямую: для плохо обусловленных матриц они могут различаться на порядки, что ещё раз подчёркивает важность анализа обусловленности.
